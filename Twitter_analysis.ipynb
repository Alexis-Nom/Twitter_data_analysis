To identify the trends on Twitter between 2015 and 2019, we approximated the evolution for all the 39 words with a linear regression. Yet some evolutions were not linear. Thus we checked the validity of the linear model with two filters :
- the coefficient of determination (R<sup>2</sup>) indicates us how well each sample of our dataset respect our linear model (0 indicating no correspondance and 1 a perfect fit)
- the p-value indicates us the siginificance of our result, e.g how likely our event happened by chance (0 indicates that our event is significant and 1 that it is the result of randomness). To get this p-value, we runned a permutation test on the slope of the linear regression with the 120 possible permutations (having a 5 elements dataset (2015-2019), there are 5! = 120 different permutations). The permutation test is a non-parametric test (no hypothesis about the distribution) which consists in shuffling the dataset (following the permutations), then computing each time the slope, and count how many times the slope is as extreme as the observed one.

So, an evolution was kept if it passes these two filters :
- R<sup>2</sup> is over 0.7, this ensures a strong fit to the linear model
- the p-value is under 0.05, considering the limitation in size of our dataset, this threshold highly decrease the risk of type I error (false postives), yet it is less efficient in preventing type II error (false negatives)

To complete our analysis, we performed a bootstraping providing us confidence intervals (CIs). Bootstraping consists in randomly resampling with replacement our original dataset to create a new distribution. Then, we take the 2.5<sup>th</sup> and 97.5<sup>th</sup> percentile of this distribution to get the lower and upper bound of our 95 % CI. Concretely, the 95 % CI means that 95 % of the time, the value will be in this interval.
